# Phase-1: Data Ingestion (London-final-light)

**This model reuses the Phase 1 outputs already produced by the main London model.**

london-final-light is an ablation of the London model that **excludes `stop_search_rate` and `seasonal_volatility`** as features. Phases 1 and 2 are identical to the main London model, so rather than recompute the expensive ingestion (NetCDF temperature, TfL footfall geocoding, spatial joins) we read the existing parquets in `outputs/phase1/`. The model only diverges from Phase 3 onward.

Data source: `outputs/phase1/`

```python
# Importing Libraries
import pandas as pd
from pathlib import Path

# Loading Datase
BASE = Path('Dataset path')
P1   = BASE / 'london_midterm' / 'outputs' / 'phase1'   # shared with the main London model

expected = [
    'phase1_crimes_london.parquet',
    'phase1_outcomes_london.parquet',
    'phase1_stop_search_london.parquet',
    'phase1_footfall_monthly.parquet',
    'phase1_temperature_monthly.parquet',
]

for fname in expected:
    f = P1 / fname
    if f.exists():
        size_mb = f.stat().st_size / 1e6
        print(f'  OK       {fname:<40} ({size_mb:6.1f} MB)')
    else:
        print(f'  MISSING  {fname} — run the main London phase1_ingestion notebook first')

print('\nPhase 1 data source: outputs/phase1/  (reused, no recomputation needed)')
```

```python
# Quick sanity check on the reused crime data
crimes = pd.read_parquet(P1 / 'phase1_crimes_london.parquet')
print(f'Street crimes (London): {crimes.shape[0]:,} rows x {crimes.shape[1]} cols')
print(f'Date range: {crimes["Month"].min():%b %Y} → {crimes["Month"].max():%b %Y}')
print(f'Forces: {sorted(crimes["Falls within"].unique())}')
```
