# Phase-2: Feature Engineering (London-final-light)

**This model reuses the Phase 2 feature matrix from the main London model.**

The shared feature matrix (`outputs/phase2/phase2_feature_matrix.parquet`) already contains every engineered feature including `stop_search_rate` and `seasonal_volatility`. London-final-light simply **does not select those two columns** when it validates and builds the index (from Phase 3 onward). No recomputation is needed.

Data source: `outputs/phase2/`

```python
# Importing Libraries
import pandas as pd
from pathlib import Path

# Loading dataset
BASE = Path('Dataset path')
P2   = BASE / 'london_midterm' / 'outputs' / 'phase2'   # shared with the main London model

fm = pd.read_parquet(P2 / 'phase2_feature_matrix.parquet')
print(f'Feature matrix shape: {fm.shape}')
print(f'Columns: {list(fm.columns)}')
print()
EXCLUDED = ['stop_search_rate', 'seasonal_volatility']
print('Columns present but EXCLUDED from london-final-light:')
for c in EXCLUDED:
    print(f'  - {c}  (present in matrix: {c in fm.columns})')
print('\nThese are dropped from FEATURES in Phase 3 onward.')
```
