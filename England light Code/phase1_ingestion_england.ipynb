# Phase-1: Data Ingestion (England_final_light)

**No re-processing needed.** Phase 1 data is identical to the England model.

This notebook simply verifies that the England Phase 1 parquets exist and reports their shape.
All downstream notebooks in this folder read directly from `England/outputs/phase1/`.

```python
# Importing Libraries 
import pandas as pd
from pathlib import Path

#Loading dataset

BASE = Path('Dataset path')
P1   = BASE / 'England' / 'outputs' / 'phase1'

for fname in ['phase1_crimes_england.parquet', 'phase1_outcomes_england.parquet']:
    f = P1 / fname
    if f.exists():
        size_mb = f.stat().st_size / 1e6
        print(f'  OK  {fname}  ({size_mb:.0f} MB)')
    else:
        print(f'  MISSING  {fname} — run England/phase1_ingestion.ipynb first')

print('\nPhase 1 data source: England/outputs/phase1/')
```
